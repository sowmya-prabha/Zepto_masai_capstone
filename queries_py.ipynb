```python
query5 = """
SELECT
    books.title,
    categories.category_name,
    books.rating,
    books.price_gbp,
    books.price_inr
FROM books
JOIN categories
    ON books.category_id = categories.category_id
ORDER BY books.rating DESC, books.price_gbp DESC
LIMIT 10
"""
```
